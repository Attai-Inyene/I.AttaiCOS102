```python
# collecting user input
print("ANNUAL TAX REVENUE CALCULATOR")
xp = int(input("Enter your years of experience: "))
age = int(input("Enter your age: "))

#declaring if statements
if xp > "25" and age >= 55:
    print("Your annual tax revenue is $5,600,000")

elif xp > "20" and age >= 45:
    print("Your annual tax revenue is $4,480,000")

elif xp > "!0" and age >= 35:
    print("Your annual tax revenue is $1,500,000")

elif xp < "10" and age < 35:
    print("Your annual tax revenue is $550,000")

else:
    print("fail to find category, try again later, sorry..")
```
