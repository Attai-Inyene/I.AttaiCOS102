```python
import math

def financial_calculator():
    print("--- Financial Calculator ---")
    print("1. Simple Interest")
    print("2. Compound Interest")
    print("3. Annuity Plan")
    choice = input("Select a calculation (1/2/3): ")

    if choice == '1':
        # Simple Interest: A = P(1 + RT/100)
        p = float(input("Enter Principal (P): "))
        r = float(input("Enter Rate (R%): "))
        t = float(input("Enter Time in years (T): "))
        amount = p * (1 + (r / 100) * t)
        print(f"Total Amount (A): {amount:.2f}")

    elif choice == '2':
        # Compound Interest: A = P(1 + r/n)^(nt)
        p = float(input("Enter Principal (P): "))
        r = float(input("Enter Annual Interest Rate (as decimal, e.g., 0.05 for 5%): "))
        n = int(input("Number of times interest compounded per year (n): "))
        t = float(input("Number of years (t): "))
        amount = p * (1 + r/n)**(n * t)
        print(f"Total Amount (A): {amount:.2f}")

    elif choice == '3':
        # Annuity Plan: A = PMT * [((1 + r/n)^(nt) - 1) / (r/n)]
        pmt = float(input("Enter Periodic Payment (PMT): "))
        r = float(input("Enter Annual Interest Rate (as decimal, e.g., 0.05): "))
        n = int(input("Number of payments per year (n): "))
        t = float(input("Number of years (t): "))
        
        # Breaking down the formula for clarity
        rate_per_period = r / n
        total_periods = n * t
        amount = pmt * (((1 + rate_per_period)**total_periods - 1) / rate_per_period)
        print(f"Future Value of Annuity (A): {amount:.2f}")

    else:
        print("Invalid selection.")

financial_calculator()
```
