```python
import tkinter as tk
from tkinter import messagebox
import pandas as pd

# 1. Load the Employee Data
try:
    # Make sure the CSV file is in the same folder as this script
    df = pd.read_csv('GIG-logistics(in).csv')
except FileNotFoundError:
    print("Error: The CSV file was not found. Please check the file name.")
    exit()

def verify_employee():
    # Get user input and remove extra spaces
    user_name = name_entry.get().strip()
    user_dept = dept_entry.get().strip()

    # Search logic: Check if Name AND Dept match (ignoring case)
    match = df[(df['Name'].str.lower() == user_name.lower()) & 
               (df['Department'].str.lower() == user_dept.lower())]

    if not match.empty:
        # Get the official name and role from the file
        official_name = match.iloc[0]['Name']
        role = match.iloc[0]['Role']
        show_success(official_name, user_dept, role)
    else:
        # The "Polite" rejection message
        messagebox.showinfo("Verification", 
                            f"We apologize, but we couldn't find an employee named '{user_name}' "
                            f"in the {user_dept} department. Please verify your details.")

def show_success(name, dept, role):
    # Create the welcome window
    top = tk.Toplevel(root)
    top.title("Employee Portal")
    top.geometry("450x400")

    # Header
    tk.Label(top, text=f"Welcome, {name}!", font=("Arial", 14, "bold"), fg="darkgreen").pack(pady=10)
    tk.Label(top, text=f"Position: {role}", font=("Arial", 10, "italic")).pack()
    
    # Department List Section
    tk.Label(top, text="\nMembers of your department:", font=("Arial", 11, "bold")).pack()
    
    # Filter the list to show colleagues (everyone in the dept except the user)
    colleagues = df[(df['Department'].str.lower() == dept.lower()) & (df['Name'] != name)]['Name'].tolist()
    
    # Put colleagues in a scrollable listbox
    list_frame = tk.Frame(top)
    list_frame.pack(pady=10, padx=20, fill="both", expand=True)
    
    scrollbar = tk.Scrollbar(list_frame)
    scrollbar.pack(side="right", fill="y")
    
    lb = tk.Listbox(list_frame, yscrollcommand=scrollbar.set, font=("Arial", 10))
    for person in colleagues:
        lb.insert("end", f" • {person}")
    
    lb.pack(side="left", fill="both", expand=True)
    scrollbar.config(command=lb.yview)

# --- MAIN LOGIN UI ---
root = tk.Tk()
root.title("GIG Logistics Verification")
root.geometry("350x300")

# UI Elements
tk.Label(root, text="GIG LOGISTICS", font=("Arial", 16, "bold")).pack(pady=10)

tk.Label(root, text="Enter Full Name:").pack()
name_entry = tk.Entry(root, width=30)
name_entry.pack(pady=5)

tk.Label(root, text="Enter Department:").pack()
dept_entry = tk.Entry(root, width=30)
dept_entry.pack(pady=5)

tk.Button(root, text="Verify Identity", command=verify_employee, bg="blue", fg="white", pady=5).pack(pady=20)

root.mainloop()
```
